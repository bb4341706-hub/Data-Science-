```python
def my_atoi(s):
    INT_MIN = -(2**31)
    INT_MAX = 2**31 - 1

    i = 0
    n = len(s)

    # Ignore leading spaces.
    while i < n and s[i] == " ":
        i += 1

    # Read the optional sign.
    sign = 1
    if i < n and (s[i] == "+" or s[i] == "-"):
        if s[i] == "-":
            sign = -1
        i += 1

    # Read digits, stopping at the first non-digit.
    value = 0
    while i < n and "0" <= s[i] <= "9":
        digit = ord(s[i]) - ord("0")

        # Clamp before adding the next digit.
        if value > INT_MAX // 10 or (
            value == INT_MAX // 10 and digit > (7 if sign == 1 else 8)
        ):
            return INT_MAX if sign == 1 else INT_MIN

        value = value * 10 + digit
        i += 1

    return sign * value


s = input("Enter a string: ")
print("Converted integer:", my_atoi(s))
```
